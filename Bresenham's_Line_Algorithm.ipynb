### Bresenham's Line Algorithm

This algorithm draws a line between two points (x0, y0) and (x1, y1) on a grid by incrementally determining the next pixel to plot based on a decision parameter. It's efficient because it avoids floating-point arithmetic.


```python
import matplotlib.pyplot as plt

def bresenham_line(x0, y0, x1, y1):
    """
    Implementation of Bresenham's Line Algorithm.
    Returns a list of (x, y) tuples representing the pixels on the line.
    """
    points = []
    dx = abs(x1 - x0)
    dy = abs(y1 - y0)
    sx = 1 if x0 < x1 else -1
    sy = 1 if y0 < y1 else -1
    err = dx - dy

    while True:
        points.append((x0, y0))
        if x0 == x1 and y0 == y1:
            break
        e2 = 2 * err
        if e2 > -dy:
            err -= dy
            x0 += sx
        if e2 < dx:
            err += dx
            y0 += sy
    return points

# Example Usage:
start_point = (2, 2)
end_point = (5, 9)
line_pixels = bresenham_line(start_point[0], start_point[1], end_point[0], end_point[1])

# Extract x and y coordinates for plotting
x_coords = [p[0] for p in line_pixels]
y_coords = [p[1] for p in line_pixels]

# Plotting the line
plt.figure(figsize=(8, 6))
plt.plot(x_coords, y_coords, 's-', color='blue', markersize=8, linewidth=2, label='Bresenham Line')
plt.scatter(x_coords, y_coords, color='red', s=50, zorder=5, label='Pixels')

plt.title(f"Bresenham's Line from {start_point} to {end_point}")
plt.xlabel("X-coordinate")
plt.ylabel("Y-coordinate")
plt.grid(True)
plt.xticks(np.arange(min(x_coords)-1, max(x_coords)+2, 1))
plt.yticks(np.arange(min(y_coords)-1, max(y_coords)+2, 1))
plt.gca().set_aspect('equal', adjustable='box')
plt.legend()
plt.show()

print(f"Generated pixels for the line: {line_pixels}")
```
